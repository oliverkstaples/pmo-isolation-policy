# Future work

The four avenues below extend `current_work.ipynb` in different directions. The first two
revisit the transmission model itself, covering which viral-kinetics model to fit and what
else drives real-world transmission heterogeneity beyond viral load. The second two turn from
the model to policy, covering how detection and compliance actually vary between people and
what happens once the isolation policy itself is a random process rather than a fixed rule.

This notebook has no code. Each section lays out the open question, why it matters, and the
literature it builds on.


## Avenue 1: model selection and fitting

Everything in this project, from the individual and hierarchical Bayesian fits through to the
isolation-policy analysis, runs on one particular target-cell-limited viral-kinetics model.
It's a reasonable choice, but it isn't the only one. The literature has several structurally
different models of comparable complexity. Zhang et al. (2023), Owens et al. (2024), and Lee
et al. (2026) are recent examples, each making different assumptions about which compartments
matter and how they interact.

The open question is whether the optimal isolation policy is sensitive to that choice. Does
fitting a more or less complex viral-kinetics model change the recommended $Q$, and can that
sensitivity be checked against independent datasets rather than just against this project's
own NBA cohort?

If model choice turns out to matter a lot, that's an important caveat on everything downstream
of it, including the isolation-policy recommendations in `current_work.ipynb`. If it doesn't,
that's reassuring: it would mean the policy conclusions are robust to a modelling choice that,
on its face, looks like it should matter.


## Avenue 2: refining infectivity itself

Right now, all the heterogeneity in transmission potential in this model is biological. It
comes entirely from between-patient variation in viral-load trajectories, through $\gamma_i$,
$\delta_i$, $\xi_i$, and $V_{0,i}$. Nothing in the model represents behavioural heterogeneity:
who actually has more contacts, or riskier ones, regardless of how much virus they're
shedding.

The broader literature suggests that's a real gap. Wegehaupt et al. (2023) on superspreading
and overdispersion, Kissler's (2026) work on bursty infectiousness, and Chen et al. (2022) all
point the same way: a substantial share of real-world transmission heterogeneity is
behavioural, driven by concentrated and variably-timed contact events, not just by
concentrated viral shedding.

The natural extension is an individually drawn contact-rate multiplier layered on top of the
viral-load-driven infectiousness this model already has. This is the same gap flagged in
`current_work.ipynb`'s Summary, where every simulated patient currently carries equal weight
in the fixed point. Drawing that multiplier from something like a Gamma distribution, with a
dispersion parameter controlling how concentrated contact activity is, would let the model ask
a genuinely new question: how does that additional, behavioural source of heterogeneity
interact with isolation policy? A highly dispersed contact-rate distribution could change
which patients matter most for outbreak risk, independent of anything the viral-kinetics
model alone can capture.


## Avenue 3: detection timing and compliance

The first two avenues are about the transmission model itself. This one, and the next, turn
from the model to how isolation policy is actually implemented and enforced.

Every analysis in `current_work.ipynb` assumes an exact $Q$-day isolation window, applied
uniformly to everyone of a given vaccination status, starting from a detection time that
coincides exactly with symptom onset ($\tau_{\text{det}}=0$, one of the locked assumptions
listed in that notebook's Summary).

In practice, detection timing relative to symptom onset varies a great deal between
individuals. It depends on testing modality, testing frequency, and how promptly someone gets
tested after developing symptoms, none of which is uniform across a population. Jeong et al.
(2022), Diao (2025), and Viloria Winnett et al. (2023) all show how sensitive
isolation-guideline design is to exactly this: testing modality and timing.

The open question is how heterogeneity in when people are actually detected, combined with
heterogeneity in how well they comply once detected, changes the residual transmission risk
this project has been computing throughout. Both $\tau_{\text{det}}$ and $\varphi$ are already
named parameters in the model. This extension would mean drawing them from distributions
instead of fixing them at $0$ and $1$, and checking whether the isolation-duration
recommendations from `current_work.ipynb` still hold once detection and compliance are
allowed to be as messy as they actually are.


## Avenue 4: flexible isolation policies

The whole PMO construction in `current_work.ipynb` assumes a fixed, vaccination-status-based
isolation policy, applied identically across every generation of the branching process. Real
contact-tracing and isolation systems aren't like that. They're stochastic and heterogeneous
by nature: tracing delays vary, some contacts are never found, and the isolation period
actually served rarely matches the period prescribed.

Extending to an inhomogeneous controlled branching process would let the model represent that
directly, drawing isolation periods and contact-tracing delays from distributions rather than
fixing them at a single value per generation. This builds on the theoretical framework in
González et al. (2015) and the classic feasibility analysis in Hellewell et al. (2020), with
roots going back to multitype branching-process theory in Allen (2015).

The open question this would let us ask: what does total excess burden look like once the
policy itself is a random variable rather than a fixed rule, instead of the deterministic $Q$
this project has swept throughout? This is the most structurally different of the four
avenues. It doesn't just add a parameter to the existing fixed point; it changes the branching
process the fixed point is built on.
